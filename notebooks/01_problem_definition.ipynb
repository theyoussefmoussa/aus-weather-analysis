# 🌦️ Australian Weather Prediction

## Problem Definition

This project aims to predict whether it will rain tomorrow in Australia based on historical weather observations using Machine Learning techniques.

## Business Problem

Accurate weather prediction helps improve decision-making in several sectors such as:

- Agriculture
- Transportation
- Disaster Management
- Water Resource Planning

## Machine Learning Problem

This is a Binary Classification problem.

Target Variable:
- RainTomorrow

Input:
- Historical weather observations.

Output:
- Yes / No

## Dataset

The dataset consists of two files:

- Weather Training Data
- Weather Test Data

The training dataset contains the target variable (RainTomorrow), while the test dataset is used for final prediction.
